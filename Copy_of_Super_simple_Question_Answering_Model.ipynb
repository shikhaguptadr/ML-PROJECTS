## **Super simple Question Answering Model**

### **Loading Transformers and pipeline**

### **Loading Context or string dataset**

```python
!   pip install PyPDF2
from transformers import pipeline

qa_model = pipeline("question-answering", "timpal0l/mdeberta-v3-base-squad2")
```

```python
from PyPDF2 import PdfReader
```

```python
path='/content/To Principal.pdf'
```

```python
pdf_file_object = open(path, 'rb')
```

```python
pdf_reader = PdfReader(pdf_file_object)
```

```python
num_pages = len(pdf_reader.pages)
print (num_pages)
```

```python
print(pdf_reader)
```

```python
extracted_text = ""
for page_num in range(num_pages):
        page_object = pdf_reader.pages[page_num]
        extracted_text += page_object.extract_text()
print(extracted_text)
```

```python
context = extracted_text
```

```python
question = "what is my name?"
qa_model(question = question, context = context)
```

```python
question = "Where do I live?"
qa_model(question = question, context = context)
```

```python
question = "what is my second home?"
qa_model(question = question, context = context)
```
